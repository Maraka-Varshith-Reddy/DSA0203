```python
#EXPERIMENT 11
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

rows, cols = img.shape[:2]

pts1 = np.float32([[50,50], [200,50], [50,200]])
pts2 = np.float32([[10,100], [200,50], [100,250]])

matrix = cv2.getAffineTransform(pts1, pts2)

result = cv2.warpAffine(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Affine Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Affine_Image.jpg", result)

print("Image saved successfully")
```

```python

#EXPERIMENT 12
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

rows, cols = img.shape[:2]

pts1 = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
pts2 = np.float32([[50,50], [cols-50,0], [0,rows-50], [cols-50,rows-50]])

matrix = cv2.getPerspectiveTransform(pts1, pts2)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Perspective Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Perspective_Image.jpg", result)

print("Image saved successfully")
```

```python

#EXPERIMENT 13
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

rows, cols = img.shape[:2]

pts1 = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
pts2 = np.float32([[0,0], [cols-1,0], [0,int(0.7*rows)], [cols-1,int(0.7*rows)]])

matrix, _ = cv2.findHomography(pts1, pts2)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Homography Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/Homography_Image.jpg", result)

print("Image saved successfully")
```

```python

#EXPERIMENT 14
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

rows, cols = img.shape[:2]

src = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
dst = np.float32([[50,50], [cols-50,20], [20,rows-50], [cols-30,rows-30]])

matrix, _ = cv2.findHomography(src, dst)

result = cv2.warpPerspective(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("DLT Transformed Image")
cv2_imshow(result)

cv2.imwrite("/content/DLT_Transformed_Image.jpg", result)

print("Image saved successfully")
```

```python
#EXPERIMENT 16
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

edges = cv2.Canny(img, 100, 200)

print("Original Image")
cv2_imshow(img)

print("Edge Detected Image")
cv2_imshow(edges)

cv2.imwrite("/content/Edges.jpg", edges)

print("Image saved successfully")
```

```python
#EXPERIMENT 17
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("sloka.jpg")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sobel_x = cv2.Sobel(img, cv2.CV_8U, 1, 0, ksize=5)

print("Original Image")
cv2_imshow(img)

print("Sobel X Image")
cv2_imshow(sobel_x)

cv2.imwrite("/content/Sobel_X.jpg", sobel_x)

print("Image saved successfully")
```
