```python
import cv2
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

img_path = "lord ganesh.jpg"
img_cv2 = cv2.imread(img_path)

img_pil = Image.fromarray(cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB))
im2 = img_pil.filter(ImageFilter.UnsharpMask(radius=3, percent=200, threshold=5))

plt.imshow(im2)
plt.axis('off')
plt.title('Processed Image (Unsharp Mask)')
plt.show()
```
