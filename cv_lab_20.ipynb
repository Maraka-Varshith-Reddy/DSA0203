```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
img = cv2.imread("lord ganesh.jpg")
kernel = np.array([[0,1,0], [1,-4,1], [0,1,0]])
sharpened = cv2.filter2D(img, -1, kernel)
cv2_imshow(sharpened)
```
